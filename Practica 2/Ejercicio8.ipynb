# Ejercicio 8 — Creación e inicialización de Arrays

```scala
val puntuaciones = new Array[Int](4)

println(puntuaciones.mkString(", "))

puntuaciones(0)= 17
puntuaciones(1)= 24
puntuaciones(2)= 21
puntuaciones(3)= 19

println(puntuaciones.mkString(", "))

println(puntuaciones.length)
```

### Documentación Ejercicio 8
- **Valores iniciales:** Un `Array[Int]` recién creado con `new Array[Int](4)` contiene ceros (`0`) por defecto en cada posición.
- **Asignación y Longitud:** Se rellenan los índices de forma manual y se obtiene el tamaño total con `.length`.

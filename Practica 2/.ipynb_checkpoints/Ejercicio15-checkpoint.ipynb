# Ejercicio 15 — Programa integrado: torneo de Twenty-One

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)
val manosRonda2 = Array(22, 19, 20, 21)

def bust(puntuacion: Int): Boolean = {
    if (puntuacion > 21) true else false
}

println("--- RONDA 1 ---")
var i = 0
while (i < jugadores.length) {
    val puntuacion = manos(i)
    
    if (bust(puntuacion)) {
        println(jugadores(i) + " con " + puntuacion + " Eliminado")
    } else if (puntuacion == 21) {
        println(jugadores(i) + " con " + puntuacion + " Puntuación perfecta ")
    } else {
        println(jugadores(i) + " con " + puntuacion + " Puntuación valida")
    }
    
    i += 1
}

println("\n--- RONDA 2 ---")
var j = 0
while (j < jugadores.length) {
    val puntuacion2 = manosRonda2(j)
    
    if (bust(puntuacion2)) {
        println(jugadores(j) + " con " + puntuacion2 + " Eliminado")
    } else if (puntuacion2 == 21) {
        println(jugadores(j) + " con " + puntuacion2 + " Puntuación perfecta ")
    } else {
        println(jugadores(j) + " con " + puntuacion2 + " Puntuación valida")
    }
    
    j += 1
}
```

### Documentación Ejercicio 15

- **Qué partes de tu solución son mutables:** Las variables de control (`i` y `j`) utilizadas para avanzar en los bucles `while`.
- **Qué partes son inmutables:** La lista de `jugadores` y los arrays de puntuaciones iniciales (`manos` y `manosRonda2`), que no se modifican en ningún momento.
- **Dónde utilizas funciones:** En la definición y llamada de la función `bust` para evaluar si una puntuación supera 21.
- **Dónde utilizas estructuras de control:** En los bucles `while` para recorrer las colecciones y en las estructuras condicionales `if / else if / else` para evaluar los estados.
- **Qué parte consideras más cercana al estilo funcional:** El uso de funciones puras independientes como `bust` para aislar la lógica de validación.
- **Qué parte consideras más cercana al estilo imperativo:** El uso de bucles `while` con variables contadoras mutables (`i`, `j`) para controlar el flujo de ejecución paso a paso.

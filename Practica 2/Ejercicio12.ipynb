# Ejercicio 12 — Operadores relacionales y lógicos

```scala
val handA = 18
val handB = 21
val handC = 25

println("HandA es mayor que HandB "+(handA>handB))
println("HandB es igual a 21: " + (handB == 21))
println("HandC es diferente de 21: " + (handC != 21))
println("Son HandA y HandB menores o iguales a 21? "+((handA<=21)&&(handB<=21)))
println("Alguna de las manos HandA o HandC supera 21: " + ((handA > 21) || (handC > 21)))
println("No se ha pasado HandB de 21: " + (!(handB > 21)))
```

### Documentación Ejercicio 12

| Expresión | Resultado | Explicación |
| :--- | :--- | :--- |
| `handA > handB` | `false` | Compara si 18 es mayor que 21. |
| `handB == 21` | `true` | Comprueba si 21 es exactamente igual a 21. |
| `handC != 21` | `true` | Valida que 25 es diferente de 21. |
| `(handA <= 21) && (handB <= 21)` | `true` | Evalúa que ambas condiciones menores o iguales a 21 se cumplan simultáneamente. |
| `(handA > 21) || (handC > 21)` | `true` | Comprueba si al menos una de las dos manos supera 21. |
| `!(handB > 21)` | `true` | Niega la condición de que `handB` sea mayor que 21. |

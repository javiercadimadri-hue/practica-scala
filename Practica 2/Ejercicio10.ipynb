# Ejercicio 10 — Listas e inmutabilidad

```scala
val jugadores = List("Alex", "Chen", "Marta")
println(jugadores.mkString(", "))
val nuevosjugadores = "Sindhu"::jugadores
println(nuevosjugadores.mkString(", "))
println(jugadores.mkString(", "))
println(jugadores.length)
println(jugadores.reverse.mkString(", "))
```

### Documentación Ejercicio 10

- **Inmutabilidad:** Las listas (`List`) son estructuras inmutables por defecto y no pueden modificarse tras su creación.
- **Operador `::`:** Crea y devuelve una nueva instancia añadiendo el elemento al principio sin alterar la original.
- **Preservación:** La lista original `jugadores` permanece completamente intacta tras realizar la operación.

# Ejercicio 11 — Construcción y concatenación de listas

```scala
val lista1 = "Ana" :: "Luis" :: "Marta" :: Nil
val lista2 = "Pedro" :: "Sofia" :: Nil
val listaunida= lista1 ::: lista2

println(lista1.mkString(", "))
println(lista2.mkString(", "))
println(listaunida.mkString(", "))
```

### Documentación Ejercicio 11

- **Construcción con `Nil` y `::`:** Permite crear listas de forma manual enlazando elementos desde el principio hasta el finalizador `Nil`.
- **Operador de concatenación `:::`:** Permite unir dos listas diferentes generando una nueva lista combinada.
- **Inmutabilidad:** Las listas originales (`lista1` y `lista2`) permanecen intactas tras la concatenación, demostrando el comportamiento seguro y sin efectos secundarios de las colecciones inmutables en Scala.

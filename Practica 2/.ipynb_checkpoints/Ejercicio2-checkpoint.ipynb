# Ejercicio 2 — val, var y reasignación

En este ejercicio se declaran variables utilizando `val` (inmutable) y `var` (mutable), se realizan operaciones de incremento sobre la puntuación, y se prueba de forma intencionada la reasignación sobre una variable inmutable para comprobar el error que genera Scala.

```scala
val jugador = "Marta"
var puntuacion = 10

puntuacion = puntuacion + 5

puntuacion = puntuacion + 3

println("Jugador: " + jugador)
println("Puntuación final: " + puntuacion)

jugador = "Lucía"
```

### Documentación Ejercicio 2
- **`puntuacion` modificable:** Declarada con `var` (permite cambios dinámicos).
- **`jugador` inmutable:** Declarada con `val` (no permite reasignaciones).
- **Error de Scala:** `reassignment to val` (prohibido modificar constantes).

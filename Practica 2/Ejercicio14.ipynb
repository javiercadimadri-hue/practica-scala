# Ejercicio 14 — Efectos secundarios y estilo de programación

```scala
var total = 0

def sumarAlTotal(valor: Int) = {
    total = total + valor
}

// 1 y 2. Ejecutar la función varias veces y mostrar el total
sumarAlTotal(5)
println("Total después de sumar 5: " + total)

sumarAlTotal(10)
println("Total después de sumar 10: " + total)

sumarAlTotal(3)
println("Total después de sumar 3: " + total)

println("-----------------------------------")

def sumar(a: Int, b: Int): Int = {
  a+b
}

println("Segunda suma: "+sumar(5,3))
```

### Documentación Ejercicio 14

| Característica | `sumarAlTotal` | `sumar` |
| :--- | :--- | :--- |
| **Modifica datos externos** | Sí (modifica la variable global `total`) | No |
| **Devuelve un resultado calculado** | No (devuelve `Unit`) | Sí (devuelve un `Int`) |
| **Utiliza efecto secundario** | Sí (altera el estado fuera de su ámbito) | No (es una función pura) |
| **Estilo predominante** | Imperativo / Procedural | Funcional |

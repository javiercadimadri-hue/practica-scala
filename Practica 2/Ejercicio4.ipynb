# Ejercicio 4 — Función para determinar si una mano se pasa de 21

```scala
def bust(puntuacion: Int): Boolean = {
    if (puntuacion > 21) {
        true
    } else {
        false
    }
}
println(bust(18))
println(bust(21))
println(bust(22))
println(bust(30))
```

### Documentación Ejercicio 4
- **Función `bust`:** Usa una estructura condicional `if/else` para comprobar si la puntuación supera 21 y retornar `true` o `false`.
- **Pureza:** Evalúa el parámetro de entrada de forma aislada sin modificar variables externas.

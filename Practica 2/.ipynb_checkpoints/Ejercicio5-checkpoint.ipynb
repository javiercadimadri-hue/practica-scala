# Ejercicio 5 — Comparación de dos manos

```scala
def maxHand(mano1: Int , mano2: Int):Int={
    if(mano1>mano2){
        mano1
    }else{
        mano2
    }
}


println(maxHand(17, 19))
println(maxHand(20, 18))
println(maxHand(21, 21))
```

### Documentación Ejercicio 5
- **Igualdad:** Si ambas manos tienen el mismo valor, el bloque `else` devuelve por defecto `mano2`.
- **Modificación:** No requiere cambios adicionales ya que devuelve correctamente el valor numérico en caso de empate.

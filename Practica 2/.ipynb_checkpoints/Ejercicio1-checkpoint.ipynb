# Ejercicio 1 — Variables, tipos e inferencia

En este ejercicio se declaran variables para representar la información de una partida de cartas mediante dos enfoques: el primero con tipado explícito (Versión A) y el segundo aprovechando la inferencia de tipos de Scala (Versión B).

```scala
// Versión A: Tipado explícito
val nombreJugadorA: String = "Javier Abadia" 
val partidasJugadasA: Int = 15
val puntuacionActualA: Int = 1200
val porcentajeVictoriasA: Double = 66.5
val estaActivoA: Boolean = true


println("JugadorA " + nombreJugadorA)
println("Partidas jugadas: " + partidasJugadasA)
println("Puntuación actual: " + puntuacionActualA)
println("Porcentaje de victorias: " + porcentajeVictoriasA + "%")
println("Está activo: " + estaActivoA)

// Versión B: Inferencia de tipos
val nombreJugadorB = "Javier Abadia"
val partidasJugadasB = 15
val puntuacionActualB = 1200
val porcentajeVictoriasB = 66.5
val estaActivoB = true


println(s"Jugador: $nombreJugadorB")
println(s"Partidas jugadas: $partidasJugadasB")
println(s"Puntuación: $puntuacionActualB")
println(s"Porcentaje de victorias: $porcentajeVictoriasB%")
println(s"Activo: $estaActivoB")
```

### Documentación Ejercicio 1
- **Tipos inferidos:** `String`, `Int`, `Double`, `Boolean`.
- **Diferencia:** Tipado explícito (`: Tipo`) en Versión A vs. deducción automática en Versión B.
- **`val` vs `var`:** `val` para constantes (`nombreJugador`); `var` para datos dinámicos (`puntuacion`, `partidas`).

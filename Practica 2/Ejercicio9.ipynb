# Ejercicio 9 — Recorrer un Array con while

```scala
def bust(puntuacion: Int): Boolean = {
    if (puntuacion > 21) {
        true
    } else {
        false
    }
}

val manos = Array(17,24,21,19,26)
var i = 0
while (i < manos.length){
    println("Mano: "+manos(i)+" | Bust: "+bust(manos(i)))
    i+=1
}
```

## Documentación — Ejercicio 9

* **Por qué el contador debe ser mutable (`var`):** Para permitir que su valor cambie e incremente de uno en uno en cada iteración del bucle.
* **Qué ocurriría si no incrementaras `i`:** Se produciría un bucle infinito y el programa se quedaría colgado porque la condición de parada nunca cambiaría.
* **Elementos de estilo imperativo:** El uso de la variable de estado mutable `var i = 0`, el control de flujo mediante el bucle `while` y la modificación manual del índice (`i += 1`).

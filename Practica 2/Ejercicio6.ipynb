# Ejercicio 6 — Decidir el ganador de una partida

```scala
def bust(puntuacion: Int): Boolean = {
    if (puntuacion > 21) {
        true
    } else {
        false
    }
}

def ganador(manoA: Int, manoB: Int): Int = {
    if (bust(manoA) && bust(manoB)) {
        0
    } else if (bust(manoA)) {
        manoB
    } else if (bust(manoB)) {
        manoA
    } else {
        if (manoA > manoB) manoA else manoB
    }
}

// Pruebas
println(ganador(26, 20))
println(ganador(18, 22))
println(ganador(24, 25))
println(ganador(17, 19))
println(ganador(21, 20))
```

### Documentación Ejercicio 6
- **`26 y 20`:** Se cumple `else if (bust(handA))` (gana `20`).
- **`18 y 22`:** Se cumple `else if (bust(handB))` (gana `18`).
- **`24 y 25`:** Se cumple el `if` inicial (`bust(handA) && bust(handB)`), devuelve `0`.
- **`17 y 19` / `21 y 20`:** Se cumple el `else` general (ninguna supera 21, gana la mayor: `19` y `21`).

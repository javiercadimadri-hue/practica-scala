# Ejercicio 7 — Arrays y mutabilidad

```scala
val jugadores = Array("Alex", "Chen", "Marta")

println(jugadores)

jugadores(0)="Sindhu"

println(jugadores)

jugadores()
jugadores(1) = 500
```

### Documentación Ejercicio 7
- **Mutabilidad:** `val` protege el contenedor, pero los elementos del `Array` pueden modificarse.
- **Error `500`:** `Array[String]` solo acepta textos, por lo que rechaza los enteros (`Int`).
- **Diferencia:** Reasignar cambia la variable entera; modificar altera solo un valor interno del array.

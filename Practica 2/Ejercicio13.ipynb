# Ejercicio 13 — foreach y funciones como valores

```scala
val manos = Array(17, 24, 21, 26, 18)

def bust(puntuacion: Int): Boolean = {
    if (puntuacion > 21) true else false
}
var i= 0
while (i<manos.length){
    println("Mano "+manos(i)+" Bust: "+ bust(manos(i)))
            i+=1
}


manos.foreach(m => println("Mano: " + m + " | Bust: " + bust(m)))
```

### Documentación Ejercicio 13

- **Uso de contador:** La **Versión A** (`while`) necesita obligatoriamente un contador numérico para rastrear la posición, mientras que la **Versión B** (`foreach`) no lo requiere.
- **Variable mutable (`var`):** La **Versión A** requiere una variable de estado mutable para avanzar el índice, a diferencia de la **Versión B** que opera de forma declarativa.
- **Estilo funcional:** La **Versión B** con `foreach` se aproxima más al estilo funcional ya que procesa los elementos de la colección pasando funciones como valores de manera abstracta.

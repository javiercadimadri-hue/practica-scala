# Ejercicio 3 — Tipos numéricos y precisión

```scala
val piDouble: Double = 3.14159265358979323846264338327
val piFloat: Float = 3.14159265358979323846264338327f

println("Double: " + piDouble)
println("Float: " + piFloat)

val alumnos: Int = 25
val aprobado: Boolean = true
val mensaje: String = "Hola"

println("Alumnos: " + alumnos)
println("Aprobado: " + aprobado)
println("Mensaje: " + mensaje)
```

### Documentación Ejercicio 3
- **`Double` vs `Float`:** `Double` ofrece mayor precisión de decimales (64 bits); `Float` tiene menor precisión (32 bits) y requiere una `f` al final.
- **Estudiantes de clase:** Tipo `Int` (números enteros).
- **Estudiante aprobado:** Tipo `Boolean` (valores verdaderos o falsos).
